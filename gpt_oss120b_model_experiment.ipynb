```python
from groq import Groq
```

```python
MODEL = "openai/gpt-oss-120b"
client = Groq()
```

### Financial Statement Construction

```python
def get_prompt(text):
    # Split by the separator line and get the last part
    parts = text.split("+----------------------------------------------------------------------------------------------------------------------+")
    # Strip leading/trailing whitespace
    result = parts[-1].strip()
    
    return result
```

```python
# Open the file in read mode
with open("prompt/income_statement.txt", "r") as file:
    # Read the entire content
    prompt_income_statement = file.read()

# Print the content
prompt_income_statement = get_prompt(prompt_income_statement)
print(prompt_income_statement)
```

```python
completion = client.chat.completions.create(
    model=MODEL,
    messages=[
      {
        "role": "user",
        "content": "I need you to construct a detailed income statement using west african style based on the following data:\n" + prompt_income_statement
      }
    ],
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

Expected output

```python
# Open the file in read mode
with open("output/income_statement.txt", "r") as file:
    # Read the entire content
    output_income_statement = file.read()

# Print the content
output_income_statement = get_prompt(output_income_statement)
print(output_income_statement)
```

### Financial Analysis

```python
# Open the file in read mode
with open("output/balance_sheet.txt", "r") as file:
    # Read the entire content
    output_balance_sheet = file.read()

# Print the content
output_balance_sheet = get_prompt(output_balance_sheet)
print(output_balance_sheet)
```

```python
# Open the file in read mode
with open("output/cash_flow_statement.txt", "r") as file:
    # Read the entire content
    output_cash_flow_statement = file.read()

# Print the content
output_cash_flow_statement = get_prompt(output_cash_flow_statement)
print(output_cash_flow_statement)
```

```python
completion = client.chat.completions.create(
    model=MODEL,
    messages=[
      {
        "role": "user",
        "content": "Make rigourious financial analysis of this Senegalese company based on the following income statement data:\n" + output_income_statement + "\n" + "balance sheet data:\n" + output_balance_sheet + "\n" + "cash flow statement data:\n" + output_cash_flow_statement
      }
    ],
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    financial_analysis = chunk.choices[0].delta.content or ""
    print(financial_analysis, end="")
```

### Investment Decision Support

```python
completion = client.chat.completions.create(
    model=MODEL,
    messages=[
      {
        "role": "user",
        "content": "Here is the financial analysis of this Senegalese company based on its income statement, balance sheet and cash flow statement:\n" + financial_analysis + "\n" + "Provide investment recommendations based on this analysis."
      }
    ],
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    investment_recommendations = chunk.choices[0].delta.content or ""
    print(investment_recommendations, end="")
```

### Risk Management

```python
completion = client.chat.completions.create(
    model=MODEL,
    messages=[
      {
        "role": "user",
        "content": "Based on the financial analysis" + financial_analysis + " and investment recommendations" + investment_recommendations + ", how this Senegalese company should deal with risks management?"
      }
    ],
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```
