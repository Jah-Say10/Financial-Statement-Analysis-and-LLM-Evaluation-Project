```python
from google import genai
from google.genai import types
```

```python
client = genai.Client()
```

```python
def get_prompt(text):
    # Split by the separator line and get the last part
    parts = text.split("+----------------------------------------------------------------------------------------------------------------------+")
    # Strip leading/trailing whitespace
    result = parts[-1].strip()
    
    return result
```

```python
# Open the file in read mode
with open("prompt/income_statement.txt", "r") as file:
    # Read the entire content
    prompt_income_statement = file.read()

# Print the content
prompt_income_statement = get_prompt(prompt_income_statement)
print(prompt_income_statement)
```

### Financial Statement Construction

```python
response = client.models.generate_content(
    model="models/gemma-3-27b-it",
    contents=["I need you to construct a detailed income statement using west african style based on the following data:\n" + prompt_income_statement],
    config=types.GenerateContentConfig(
        temperature=1
    )
)
print(response.text)
```

Expected output

```python
# Open the file in read mode
with open("output/income_statement.txt", "r") as file:
    # Read the entire content
    output_income_statement = file.read()

# Print the content
output_income_statement = get_prompt(output_income_statement)
print(output_income_statement)
```

### Financial Analysis

```python
# Open the file in read mode
with open("output/balance_sheet.txt", "r") as file:
    # Read the entire content
    output_balance_sheet = file.read()

# Print the content
output_balance_sheet = get_prompt(output_balance_sheet)
print(output_balance_sheet)
```

```python
# Open the file in read mode
with open("output/cash_flow_statement.txt", "r") as file:
    # Read the entire content
    output_cash_flow_statement = file.read()

# Print the content
output_cash_flow_statement = get_prompt(output_cash_flow_statement)
print(output_cash_flow_statement)
```

```python
response = client.models.generate_content(
    model="models/gemma-3-27b-it",
    contents=["Make rigourious financial analysis of this Senegalese company based on the following income statement data:\n" + output_income_statement + "\n" + "balance sheet data:\n" + output_balance_sheet + "\n" + "cash flow statement data:\n" + output_cash_flow_statement],
    config=types.GenerateContentConfig(
        temperature=1
    )
)

financial_analysis = response.text
print(financial_analysis)
```

### Investment Decision Support. 

```python
response = client.models.generate_content(
    model="models/gemma-3-27b-it",
    contents=["Here is the financial analysis of this Senegalese company based on its income statement, balance sheet and cash flow statement:\n" + financial_analysis + "\n" + "Provide investment recommendations based on this analysis."],
    config=types.GenerateContentConfig(
        temperature=1
    )
)

investment_recommendations = response.text
print(investment_recommendations)
```

### Risk Management

```python
response = client.models.generate_content(
    model="models/gemma-3-27b-it",
    contents=["Based on the financial analysis" + financial_analysis + " and investment recommendations" + investment_recommendations + ", how  this Senegalese company should deal with risks management?"],
    config=types.GenerateContentConfig(
        temperature=1
    )
)

riks_management = response.text
print(riks_management)
```

### Forecasting and Predictive Analytics
